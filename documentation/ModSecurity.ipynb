## **1. Pruebas para ModSecurity (WAF)**

### **1.1. Pruebas de protección contra ataques comunes**
- **SQL Injection**:
  - Enviar una solicitud con un payload SQL malicioso, como:
    ```
    http://localhost:8080/search?q=' OR '1'='1
    ```
  - Verificar que ModSecurity bloquee la solicitud y devuelva un código de estado `403 Forbidden`.

- **Cross-Site Scripting (XSS)**:
  - Enviar un formulario con un script malicioso, como:
    ```html
    <script>alert('XSS');</script>
    ```
  - Verificar que ModSecurity detecte y bloquee el payload.

- **Fuerza bruta**:
  - Intentar múltiples solicitudes de inicio de sesión con credenciales incorrectas.
  - Verificar que ModSecurity bloquee la IP después de un número configurado de intentos.

- **File Inclusion**:
  - Intentar acceder a archivos sensibles, como:
    ```
    http://localhost:8080/?page=../../etc/passwd
    ```
  - Verificar que ModSecurity bloquee la solicitud.

- **Ataques de inyección de comandos**:
  - Enviar una solicitud con un comando malicioso, como:
    ```
    http://localhost:8080/exec?cmd=rm -rf /
    ```
  - Verificar que ModSecurity bloquee la solicitud.

---

### **1.2. Pruebas de configuración de ModSecurity**
- **Reglas básicas**:
  - Verificar que las reglas básicas de OWASP CRS (Core Rule Set) estén cargadas y activas.
  - Ejecutar una solicitud maliciosa y verificar que se registre en los logs de ModSecurity.

- **Configuración de logs**:
  - Verificar que los logs de ModSecurity se almacenen correctamente en `/var/log/modsec_audit.log`.
  - Asegurarse de que los logs contengan detalles sobre las solicitudes bloqueadas.

- **Modo de detección**:
  - Configurar ModSecurity en modo de detección (solo registro, sin bloquear) y verificar que las solicitudes maliciosas se registren pero no se bloqueen.

- **Modo de bloqueo**:
  - Configurar ModSecurity en modo de bloqueo y verificar que las solicitudes maliciosas sean bloqueadas.

Para probar si tu aplicación web es vulnerable a **Cross-Site Scripting (XSS)** y verificar si **ModSecurity** está correctamente configurado para bloquear estos ataques, puedes seguir estos pasos:

---

## **1. Crear un formulario en tu aplicación web**

Primero, asegúrate de que tu aplicación web tenga un formulario que permita la entrada de datos por parte del usuario. Por ejemplo, un formulario de comentarios o un campo de búsqueda.

Aquí tienes un ejemplo básico de un formulario HTML:

```html
<!DOCTYPE html>
<html lang="es">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Formulario de prueba</title>
</head>
<body>
    <h1>Formulario de comentarios</h1>
    <form action="/submit-comment" method="POST">
        <label for="comment">Deja un comentario:</label><br>
        <textarea id="comment" name="comment" rows="4" cols="50"></textarea><br><br>
        <input type="submit" value="Enviar">
    </form>
</body>
</html>
```

---

## **2. Enviar un payload XSS**

Una vez que tengas el formulario, puedes probar si tu aplicación es vulnerable a XSS enviando un script malicioso. Por ejemplo:

```html
<script>alert('XSS');</script>
```

### **Pasos para enviar el payload XSS:**

1. Abre tu aplicación web en el navegador.
2. Completa el formulario con el siguiente contenido en el campo de comentarios:
   ```
   <script>alert('XSS');</script>
   ```
3. Envía el formulario.

---

## **3. Verificar el comportamiento de la aplicación**

Dependiendo de cómo esté configurada tu aplicación y ModSecurity, pueden ocurrir dos escenarios:

### **Escenario 1: La aplicación es vulnerable a XSS**
- Si la aplicación es vulnerable, verás una alerta en el navegador con el mensaje `XSS`. Esto significa que el script se ejecutó correctamente, lo que indica una vulnerabilidad de seguridad.

### **Escenario 2: ModSecurity bloquea el payload XSS**
- Si ModSecurity está correctamente configurado, debería bloquear la solicitud y devolver un código de estado `403 Forbidden`. Además, el script no se ejecutará en el navegador.

---

## **4. Verificar los logs de ModSecurity**

Si ModSecurity bloquea la solicitud, puedes verificar los logs para confirmar que el payload XSS fue detectado y bloqueado.

1. Accede a los logs de ModSecurity, que generalmente se encuentran en:
   ```
   /var/log/modsec_audit.log
   ```
2. Busca entradas relacionadas con la solicitud bloqueada. Deberías ver algo como:
   ```
   [msg "XSS Attack Detected"]
   [data "<script>alert('XSS');</script>"]
   ```

---

## **5. Configurar ModSecurity para bloquear XSS (si no lo está)**

Si ModSecurity no bloqueó el payload XSS, asegúrate de que las reglas de OWASP Core Rule Set (CRS) estén correctamente configuradas. Aquí te explico cómo hacerlo:

1. **Verificar que las reglas de OWASP CRS estén activas**:
   - Asegúrate de que el archivo de configuración de ModSecurity incluya las reglas de OWASP CRS. Por ejemplo:
     ```nginx
     modsecurity_rules_file /etc/nginx/modsec/crs-setup.conf;
     modsecurity_rules_file /etc/nginx/modsec/rules/*.conf;
     ```

2. **Habilitar la regla específica para XSS**:
   - Las reglas de XSS están incluidas en el archivo `REQUEST-941-APPLICATION-ATTACK-XSS.conf` de OWASP CRS. Asegúrate de que este archivo esté cargado.

3. **Reiniciar Nginx**:
   - Después de realizar cambios en la configuración de ModSecurity, reinicia Nginx para aplicar los cambios:
     ```bash
     sudo systemctl restart nginx
     ```

## **7. Mejores prácticas para prevenir XSS**

Además de usar ModSecurity, sigue estas mejores prácticas para prevenir XSS en tu aplicación:

1. **Escape de salida**:
   - Asegúrate de que todos los datos proporcionados por el usuario se escapen correctamente antes de mostrarlos en la página. Por ejemplo, en PHP puedes usar `htmlspecialchars()`.

2. **Validación de entradas**:
   - Valida y sanitiza todas las entradas del usuario antes de procesarlas.

3. **Usar Content Security Policy (CSP)**:
   - Configura una política de seguridad de contenido (CSP) para restringir la ejecución de scripts no permitidos.

4. **Usar frameworks seguros**:
   - Si estás usando un framework como React, Angular o Vue, aprovecha sus mecanismos de seguridad integrados para prevenir XSS.
