# Guía para marcar una cuenta como verificada manualmente en SQLite

Pasos para verificar una cuenta de usuario: Ejemplo: "davgalle":

## Paso 1: Conectarse a la base de datos SQLite
```bash
docker exec -it sqlite sh
sqlite3 sqlite.db
```

## Paso 2: Examinar la estructura de la tabla de usuarios
```sql
PRAGMA table_info(users);
```

Busca columna:
- `is_verified`

## Paso 3: Buscar al usuario específico
```sql
SELECT * FROM users WHERE username = 'davgalle';
```

## Paso 4: Actualizar el estado de verificación

```sql
UPDATE users SET is_verified = 1 WHERE username = 'davgalle';
```

## Paso 5: Verificar el cambio
```sql
SELECT username, verified, email_verified_at FROM users WHERE username = 'davgalle';
```

## Paso 6: Salir de SQLite
```sql
.quit
```
