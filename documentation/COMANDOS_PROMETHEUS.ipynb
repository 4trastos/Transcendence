## ✅ **Comandos de Prometheus para probar en consola (Web UI)**

En `http://localhost:9090` (la UI de Prometheus), abre la consola de consultas y prueba:

### 🔸 CPU usado por tu app (total)

```promql
process_cpu_seconds_total{job="backend"}
```

### 🔸 Memoria residente

```promql
process_resident_memory_bytes{job="backend"}
```

### 🔸 Número de file descriptors abiertos

```promql
process_open_fds{job="backend"}
```

### 🔸 Event loop lag de Node.js

```promql
nodejs_eventloop_lag_seconds{job="backend"}
```

### 🔸 Uso de heap en Node.js

```promql
nodejs_heap_size_used_bytes{job="backend"}
```

### 🔸 Métricas del GC (garbage collector)

```promql
nodejs_gc_duration_seconds_count{job="backend"}
```
